```python
import telemetry
from events import EVENT_STRUCT, Kind, decode
from opentelemetry.sdk._logs.export import InMemoryLogRecordExporter

raw = EVENT_STRUCT.pack(
    Kind.OPEN,
    42,
    7,
    1000,
    b"cat".ljust(16, b"\0"),
    b"decoys/.aws/credentials".ljust(256, b"\0"),
)
event = decode(raw)
assert event.detail == "decoys/.aws/credentials"

exporter = InMemoryLogRecordExporter()
emitter = telemetry.Emitter(telemetry.provider(exporter))
emitter.emit(event)
emitter.shutdown()
records = exporter.get_finished_logs()
assert len(records) == 1
print(dict(records[0].log_record.attributes))
```
