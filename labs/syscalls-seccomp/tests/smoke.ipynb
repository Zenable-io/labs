```python
import profile_from_strace
import seccomp_sandbox
from syscalls import number

program = seccomp_sandbox.build_filter(
    [number("connect")], seccomp_sandbox.ACTIONS["errno"]
)
assert len(program) == 8 * 9
print(seccomp_sandbox.describe({"connect": number("connect")}, "errno"))

profile = profile_from_strace.profile({"read", "write", "openat"})
assert profile["defaultAction"] == "SCMP_ACT_ERRNO"
print(len(profile["syscalls"][0]["names"]), "syscalls allowed")
```
