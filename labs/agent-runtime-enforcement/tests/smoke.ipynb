```python
import nono_agent
from nono_py import AccessMode, QueryContext

# The capability set is pure data until apply(); QueryContext answers what it
# would allow without touching the kernel, so this runs anywhere.
caps = nono_agent.capabilities()
ctx = QueryContext(caps)
here = nono_agent.HERE
assert (
    ctx.query_path(str(here / "workspace" / "VERSION"), AccessMode.READ)["status"]
    == "allowed"
)
assert (
    ctx.query_path(str(here / "decoys" / ".aws" / "credentials"), AccessMode.READ)[
        "status"
    ]
    == "denied"
)
assert ctx.query_network()["status"] == "denied"
print(caps.summary())
```
